4. TASK 1.1 — GENERATION OF BASIC DISCRETE-TIME SIGNALS

4.1 Procedure

A discrete-time index was created using:

[
n=-5,-4,\ldots,0,\ldots,10

Unit impulse signal

u(n)=

[
r(n)=nu(n)

[
x(n)=(0.8)^nu(n)

4.4 Observation

task 1.1

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)

# Generate discrete-time index
n = np.arange(-5, 11)

# Elementary signals
delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step

# Plot signals
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

signals = [delta, step, ramp, expo]
titles = ["Unit Impulse", "Unit Step",
          "Unit Ramp", "Exponential"]

for ax, x, title in zip(axes, signals, titles):
    ax.stem(n, x)
    ax.set_title(title)
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

5. TASK 1.2 — CHANGE THE PARAMETERS

5.1 Exponential signals

The exponential signal is defined as:

[
x(n)=a^nu(n)
]

Three different values of (a) were investigated.

Case 1: (a=0.5)

[
x(n)=(0.5)^nu(n)
]

The signal decreases rapidly as (n) increases.

Case 2: (a=-0.8)

[
x(n)=(-0.8)^nu(n)
]

The signal alternates between positive and negative values while its magnitude decreases.

Case 3: (a=1.1)

[
x(n)=(1.1)^nu(n)
]

The signal increases exponentially as (n) increases.

5.2 Cosine signals

The following signals were investigated:

[
x_1(n)=\cos\left(\frac{2\pi n}{8}\right)
]

[
x_2(n)=\cos\left(\frac{2\pi n}{7.5}\right)
]

A discrete-time cosine signal is periodic if its normalized frequency is a rational number.

For the first signal:

[
\frac{\omega_1}{2\pi}=\frac{1}{8}
]

Therefore, its fundamental period is:

[
\boxed{N_1=8}
]

For the second signal:

[
\frac{\omega_2}{2\pi}=\frac{1}{7.5}
=\frac{2}{15}
]

Therefore, its fundamental period is:

[
\boxed{N_2=15}
]

5.3 Observation

Both cosine signals repeat exactly. The first signal repeats after 8 samples, while the second signal repeats after 15 samples.

```python

# Part 1: Exponential signals
n = np.arange(-5, 11)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, a_values):
    x = (a**n) * (n >= 0)
    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()

# Part 2: Cosine signals
n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

plt.figure(figsize=(10, 4))
plt.stem(n, x1, linefmt='b-', markerfmt='bo',
         basefmt='k-', label='cos(2πn/8)')
plt.plot(n, x2, 'r--o', label='cos(2πn/7.5)')

plt.xlabel("n")
plt.ylabel("Amplitude")
plt.title("Comparison of Cosine Signals")
plt.legend()
plt.grid(True)
plt.show()
```

6.2 Sampling information

The sampling rate used was:

[
f_s=16,000\text{ Hz}
]

This means that the audio system records 16,000 samples per second.

The number of samples is calculated using:

[
N=f_s\times T
]

Where:

(N) = total number of samples.

(f_s) = sampling frequency.

(T) = recording duration in seconds.

For one minute of mono audio:

[
N=16,000\times60
]

[
\boxed{N=960,000\text{ samples}}
]

For the actual recording, the duration was [insert duration from Google Colab] seconds.

Therefore:

[
N=16,000\times[\text{actual duration}]
]

[

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from IPython.display import Audio, display

fs, x = wavfile.read("/content/my_voice.wav")

x = x.astype(float)

if x.ndim > 1:
    x = x.mean(axis=1)

if np.max(np.abs(x)) > 0:
    x = x / np.max(np.abs(x))

t = np.arange(len(x)) / fs

print("MY VOICE RECORDING")
print("------------------")
print(f"Sampling rate = {fs} Hz")
print(f"Duration = {len(x)/fs:.2f} seconds")
print(f"Number of samples = {len(x)}")

plt.figure(figsize=(12, 4))
plt.plot(t, x)
plt.title("My Voice Recording")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

display(Audio(x, rate=fs))
```

Question 2

Why is a stem plot more appropriate than a continuous line for (x(n))?

A stem plot is more appropriate because a discrete-time signal is defined only at integer values of (n). It clearly displays individual samples and their amplitudes without suggesting that the signal exists continuously between samples.

Question 3

For (a=1.1), the exponential grows. Which property of a system would such a signal violate if it were an impulse response?

It would violate the BIBO stability property.

For a discrete-time system to be BIBO stable, its impulse response must satisfy:

[
\sum_{n=-\infty}^{\infty}|h(n)|<\infty
]

For an impulse response of the form:

[
h(n)=(1.1)^nu(n)
]

the sum is:

[
\sum_{n=0}^{\infty}(1.1)^n
]

Since the terms grow instead of approaching zero, the sum diverges.

Therefore, the system is not BIBO stable.
